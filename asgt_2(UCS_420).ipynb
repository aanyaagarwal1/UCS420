```python
roll_number = input("Enter your roll number: ")

L = [int(digit) * 10 for digit in roll_number]

print("L =", L)

L.append(int(input("Enter a number to append: ")))
print("After append:", L)

position = int(input("Enter the position for insert: "))
number = int(input("Enter a number to insert: "))
L.insert(position, number)
print("After insert:", L)

remove_value = int(input("Enter the value to remove: "))
if remove_value in L:
    L.remove(remove_value)
print("After remove:", L)

if L:
    L.pop()
print("After pop:", L)

L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)

print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

average = sum(L) / len(L)
above_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", above_average)
```

```python
scores = tuple(L[:8])

print("Scores:", scores)

highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)

reversed_scores = list(scores[::-1])
print("Reversed tuple as list:", reversed_scores)

score = int(input("Enter a score: "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score not present")

try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)

first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
import random

random.seed(int(roll_number))

numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:", numbers)

odd_numbers = [x for x in numbers if x % 2 != 0]
even_numbers = [x for x in numbers if x % 2 == 0]

print("Number of odd numbers:", len(odd_numbers))
print("Odd numbers:", odd_numbers)

print("Number of even numbers:", len(even_numbers))
print("Even numbers:", even_numbers)

def is_prime(number):
    if number < 2:
        return False

    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print("Number of prime numbers:", len(prime_numbers))
print("Prime numbers:", prime_numbers)

most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("Most frequent number:", most_frequent)
print("Number of occurrences:", frequency)
```

```python
digits = [int(digit) for digit in roll_number]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

print("Union:", A.union(B))

print("Intersection:", A.intersection(B))

print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

print("Symmetric difference:", A.symmetric_difference(B))

print("Is A a subset of B:", A.issubset(B))
print("Is B a superset of A:", B.issuperset(A))

x = int(input("Enter a value X: "))
A.discard(x)

print("Set A after discard:", A)
```

```python
name = input("Enter your name: ")
roll_no = input("Enter your roll number: ")
branch = input("Enter your branch: ")
age = int(input("Enter your age: "))
city = input("Enter your home city: ")
cgpa = float(input("Enter your CGPA: "))

my_dict = {
    "name": name,
    "roll_no": roll_no,
    "branch": branch,
    "age": age,
    "city": city
}

print("Original dictionary:", my_dict)

my_dict["location"] = my_dict.pop("city")
print("After renaming city to location:", my_dict)

my_dict["cgpa"] = cgpa
print("After adding CGPA:", my_dict)

my_dict["age"] += 1
print("After increasing age:", my_dict)

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

dict_pop.pop("branch")
del dict_del["branch"]

print("Using pop():", dict_pop)
print("Using del:", dict_del)

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist")

friend_dict = {
    "name": "Rahul",
    "roll_no": "12345678",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:", merged_dict)

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Dictionary containing only string values:", string_values)
```
